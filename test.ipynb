```python
def clean_text(text):
    return text.replace('¸', 'ü').replace('ˆ', 'ö')

def clean_sales(df, plz_df):
    df = df.copy()

    df['PLZ'] = df['PLZ'].astype(str).str.zfill(5)
    df['BestellNr.'] = df['BestellNr.'].astype(str)
    df['Artikel'] = df['Artikel'].astype(str)
    df['PZN'] = df['PZN'].astype(str)

    df['Ort'] = df['Ort'].apply(clean_text)
    df['Bezeichnung'] = df['Bezeichnung'].apply(clean_text)

    # Clean Umsatz values
    mask = df['Umsatz'].apply(lambda x: isinstance(x, str))

    df.loc[mask, 'Umsatz'] = (
        df.loc[mask, 'Umsatz']
        .str.replace(',', '.', regex=False)
    )

    df['Umsatz'] = pd.to_numeric(
        df['Umsatz'],
        errors='coerce'
    )

    # Free deliveries have no revenue
    free_delivery_mask = (
        df['Umsatz'].isna()
        & df['Auftragsart'].eq('Kostenlose Lieferung')
    )

    df.loc[free_delivery_mask, 'Umsatz'] = 0

    df['Versanddatum'] = pd.to_datetime(
        df['Versanddatum'],
        dayfirst=True,
        errors='coerce'
    )

    df['Berichtsmonat'] = df['Versanddatum'].dt.to_period('M').astype(str)

    df = df.merge(plz_df, on='PLZ', how='left')

    return df
```

```python
def clean_inventory(df):
    df = df.copy()

    df['ArtikelNr'] = df['ArtikelNr'].astype(str)
    df['Charge'] = df['Charge'].astype(str)
    df['Bezeichnung'] = df['Bezeichnung'].apply(clean_text)

    df[['Zugang', 'Abgang', 'Retouren', 'sonstige Vernichtung', 'Inventuren']] = (
        df[['Zugang', 'Abgang', 'Retouren', 'sonstige Vernichtung', 'Inventuren']].fillna(0)
    )

    return df
```

```python
# Load clean and combine multiple Inventory files
import re
inventory_files = [
    "data/raw/inventory_2026/" + f
    for f in os.listdir("data/raw/inventory_2026")
    if f.startswith("Lagerliste")
]

data_frames = []

for file in inventory_files:
    df = pd.read_excel(file)
    df = clean_inventory(df)

    match = re.search(r'\d{4}-\d{2}-\d{2}', file)

    if match:
        df['Berichtsmonat'] = match.group(0)[:7]

    data_frames.append(df)

inventory_all = pd.concat(data_frames, ignore_index=True)
```

```python
#loop erstellen für e.g jan2025 vs jan 2026 
from datetime import datetime

from datetime import datetime

daten_vorjahr = [
    datetime(2025, 1, 15),
    datetime(2025, 2, 20),
    datetime(2025, 3, 10)
]

daten_aktuell = [
    datetime(2026, 1, 18),
    datetime(2026, 2, 22),
    datetime(2026, 4, 5)
]

for datum_alt, datum_neu in zip(daten_vorjahr, daten_aktuell):
    if datum_alt.month == datum_neu.month:
        print(f"✅ Richtig! Monat {datum_alt.month} ({datum_alt.year}) wird mit Monat {datum_neu.month} ({datum_neu.year}) verglichen.")
    else:
        print(f"❌ Fehler! Monat {datum_alt.month} ({datum_alt.year}) passt nicht zu Monat {datum_neu.month} ({datum_neu.year})!")
```

```python
# Find the same month in the previous year

def get_previous_year_month(report_month):
    year, month = report_month.split("-")
    return f"{int(year) - 1}-{month}"
```
